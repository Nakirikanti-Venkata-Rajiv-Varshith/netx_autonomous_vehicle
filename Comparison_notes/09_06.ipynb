```python
import numpy as np
import matplotlib.pyplot as plt

# --------------------------------------------------
# HELPER FUNCTIONS
# --------------------------------------------------

def annotate_bars(ax):
    for p in ax.patches:
        h = p.get_height()
        ax.annotate(
            f'{h:.1f}',
            (p.get_x() + p.get_width()/2, h),
            ha='center',
            va='bottom',
            fontsize=8,
            rotation=0
        )

def annotate_line(ax, x, y):
    for xi, yi in zip(x, y):
        ax.annotate(
            f'{yi:.1f}',
            (xi, yi),
            textcoords="offset points",
            xytext=(0, 5),
            ha='center',
            fontsize=8
        )

# --------------------------------------------------
# PLOT FUNCTION
# --------------------------------------------------

systems = ["Onboard", "ALB", "Offboard"]

def plot_condition(data, title):

    fig, axs = plt.subplots(3, 3, figsize=(20, 14))
    fig.suptitle(title, fontsize=20, fontweight='bold')

    # ==================================================
    # 1. RESOURCE UTILIZATION
    # ==================================================
    metrics = ['CPU','RAM','GPU','GPU Active']

    x = np.arange(len(metrics))
    width = 0.25

    for i, sys in enumerate(systems):

        vals = data[sys]["resource"][:4]

        bars = axs[0,0].bar(
            x + i*width,
            vals,
            width,
            label=sys
        )

        for b in bars:
            axs[0,0].text(
                b.get_x()+b.get_width()/2,
                b.get_height(),
                f'{b.get_height():.1f}',
                ha='center',
                fontsize=8
            )

    axs[0,0].set_title("Average Resource Utilization (%)")
    axs[0,0].set_xticks(x + width)
    axs[0,0].set_xticklabels(metrics)
    axs[0,0].set_ylim(0,110)
    axs[0,0].legend()

    # ==================================================
    # 2. POWER
    # ==================================================
    power_vals = [
        data[s]["resource"][4]
        for s in systems
    ]

    bars = axs[0,1].bar(
        systems,
        power_vals
    )

    annotate_bars(axs[0,1])

    axs[0,1].set_title("Power Consumption (mW)")

    # ==================================================
    # 3. FPS
    # ==================================================
    fps_vals = [
        data[s]["fps"][0]
        for s in systems
    ]

    bars = axs[0,2].bar(
        systems,
        fps_vals
    )

    annotate_bars(axs[0,2])

    axs[0,2].set_title("FPS")

    # ==================================================
    # 4. FRAME LATENCY
    # ==================================================
    latency_names = ['P50','P90','P99','P99.99']

    for sys in systems:

        y = data[sys]["frame_lat"]

        axs[1,0].plot(
            latency_names,
            y,
            marker='o',
            linewidth=2,
            label=sys
        )

        annotate_line(
            axs[1,0],
            latency_names,
            y
        )

    axs[1,0].set_title("Frame Latencies (ms)")
    axs[1,0].set_yscale('log')
    axs[1,0].legend()

    # ==================================================
    # 5. OVERALL E2E
    # ==================================================
    overall_labels = ['P50','P90','P99']

    for sys in systems:

        y = data[sys]["overall"]

        axs[1,1].plot(
            overall_labels,
            y,
            marker='o',
            linewidth=2,
            label=sys
        )

        annotate_line(
            axs[1,1],
            overall_labels,
            y
        )

    axs[1,1].set_title("Overall E2E Latency (ms)")
    axs[1,1].set_yscale('log')
    axs[1,1].legend()

    # ==================================================
    # 6. APPLICATION P99 LATENCY
    # ==================================================
    apps = [
        'Collision Avoidance',
        'Lane Detection',
        'Traffic Sign'
    ]

    for sys in systems:

        vals = [
            data[sys]["apps"][a][2]
            for a in apps
        ]

        axs[1,2].plot(
            apps,
            vals,
            marker='o',
            linewidth=2,
            label=sys
        )

        annotate_line(
            axs[1,2],
            apps,
            vals
        )

    axs[1,2].set_title("Application P99 E2E Latency")
    axs[1,2].set_yscale('log')
    axs[1,2].tick_params(axis='x', rotation=15)
    axs[1,2].legend()

    # ==================================================
    # 7. ONBOARD COUNTS
    # ==================================================
    apps_short = ['CA','LD','TS']

    x = np.arange(len(apps_short))
    width = 0.25

    for i, sys in enumerate(systems):

        vals = data[sys]["onboard"]

        bars = axs[2,0].bar(
            x + i*width,
            vals,
            width,
            label=sys
        )

        for b in bars:
            axs[2,0].text(
                b.get_x()+b.get_width()/2,
                b.get_height(),
                f'{int(b.get_height())}',
                ha='center',
                fontsize=8
            )

    axs[2,0].set_title("Onboard Counts")
    axs[2,0].set_xticks(x + width)
    axs[2,0].set_xticklabels(apps_short)
    axs[2,0].legend()

    # ==================================================
    # 8. OFFLOAD COUNTS
    # ==================================================
    for i, sys in enumerate(systems):

        vals = data[sys]["offload"]

        bars = axs[2,1].bar(
            x + i*width,
            vals,
            width,
            label=sys
        )

        for b in bars:
            axs[2,1].text(
                b.get_x()+b.get_width()/2,
                b.get_height(),
                f'{int(b.get_height())}',
                ha='center',
                fontsize=8
            )

    axs[2,1].set_title("Offload Counts")
    axs[2,1].set_xticks(x + width)
    axs[2,1].set_xticklabels(apps_short)
    axs[2,1].legend()

    # ==================================================
    # 9. OFFLOAD RATIO (%)
    # ==================================================
    ratios = []

    for sys in systems:

        on = np.sum(data[sys]["onboard"])
        off = np.sum(data[sys]["offload"])

        total = on + off

        ratios.append(
            100 * off / total if total > 0 else 0
        )

    bars = axs[2,2].bar(
        systems,
        ratios
    )

    annotate_bars(axs[2,2])

    axs[2,2].set_ylim(0,100)
    axs[2,2].set_title("Offload Ratio (%)")

    plt.tight_layout()
    plt.show()


# --------------------------------------------------
# CALLS
# --------------------------------------------------

plot_condition(
    normal,
    "NORMAL CONDITIONS (Onboard → ALB → Offboard)"
)

plot_condition(
    stress,
    "STRESS CONDITIONS (Onboard → ALB → Offboard)"
)
```
