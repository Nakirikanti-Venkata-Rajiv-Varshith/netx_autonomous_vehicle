```python
import numpy as np
import matplotlib.pyplot as plt

# ============================================================
# DATA
# Order: Onboard vs ALB vs Offboard
# ============================================================

schemes = ["Onboard", "ALB", "Offboard"]
COLORS = ["#2ca02c", "#ff7f0e", "#1f77b4"]   # green, orange, blue

# ---------------- NORMAL ----------------
normal_resource = {
    "CPU":        [80.98, 71.59, 48.44],
    "RAM":        [71.70, 74.04, 64.35],
    "GPU Active": [69.22, 52.35, 1.04],
    "Power":      [5294.32, 4616.73, 3846.73]
}

normal_fps = [9.96, 10.10, 9.97]

normal_e2e = {
    "P50": [25.00, 35.80, 83.40],
    "P90": [33.20, 93.96, 101.30],
    "P99": [44.96, 172.55, 188.74]
}

normal_app_latency = {
    "Collision Avoidance": {
        "P50": [24.60, 24.90, 53.40],
        "P90": [40.60, 46.22, 71.30],
        "P99": [99.10, 101.06, 157.04]
    },
    "Lane Detection": {
        "P50": [19.20, 19.60, 53.20],
        "P90": [41.98, 44.38, 71.10],
        "P99": [89.93, 121.66, 156.84]
    },
    "Traffic Sign Detection": {
        "P50": [30.30, 83.10, 83.60],
        "P90": [59.88, 87.79, 91.62],
        "P99": [113.46, 160.17, 168.35]
    }
}

# ---------------- STRESS ----------------
stress_resource = {
    "CPU":        [95.31, 89.28, 85.08],
    "RAM":        [86.49, 89.27, 90.43],
    "GPU Active": [80.09, 63.74, 80.71],
    "Power":      [8162.99, 7781.58, 7276.99]
}

stress_fps = [8.49, 8.25, 8.78]

stress_e2e = {
    "P50": [149.30,  90.90,  115.20],
    "P90": [386.64, 271.10, 219.60],
    "P99": [1856.10, 1165.19, 692.06]
}

stress_app_latency = {
    "Collision Avoidance": {
        "P50": [148.30,  85.00,  114.95],
        "P90": [279.00, 259.80, 216.22],
        "P99": [1730.00, 1032.00, 614.83]
    },
    "Lane Detection": {
        "P50": [144.60,  83.05,  114.70],
        "P90": [267.68, 259.05, 216.03],
        "P99": [1790.01, 1031.79, 612.74]
    },
    "Traffic Sign Detection": {
        "P50": [161.00,  106.30, 116.25],
        "P90": [399.48, 304.99, 216.42],
        "P99": [1856.10, 1277.43, 615.36]
    }
}

# ============================================================
# HELPER
# ============================================================

plt.rcParams.update({
    "font.size":       11,
    "axes.titlesize":  13,
    "axes.labelsize":  11,
    "legend.fontsize": 10
})

def add_labels(ax, bars, fmt="{:.1f}"):
    for b in bars:
        h = b.get_height()
        ax.annotate(
            fmt.format(h),
            xy=(b.get_x() + b.get_width() / 2, h),
            xytext=(0, 3),
            textcoords="offset points",
            ha="center",
            fontsize=8
        )

# ============================================================
# PLOTTING FUNCTION
# ============================================================

def create_comparison_plots(title_prefix, resource_data, fps_data,
                             e2e_data, app_latency):

    # ----------------------------------------------------------
    # Graph 1 : CPU / RAM / GPU Active — grouped by METRIC
    #           Each metric cluster: Onboard | ALB | Offboard
    # ----------------------------------------------------------
    metrics   = ["CPU", "RAM", "GPU Active"]
    n_metrics = len(metrics)
    n_schemes = len(schemes)

    cluster_gap = 1.0          # gap between metric clusters
    bar_width   = 0.22
    offsets     = np.array([-bar_width, 0, bar_width])

    # x positions for cluster centres
    x_centres = np.arange(n_metrics) * (n_schemes * bar_width + cluster_gap)

    fig, ax = plt.subplots(figsize=(11, 6))

    for i, (scheme, color) in enumerate(zip(schemes, COLORS)):
        values = [resource_data[m][i] for m in metrics]
        bars = ax.bar(x_centres + offsets[i], values, bar_width,
                      label=scheme, color=color)
        add_labels(ax, bars)

    ax.set_title(f"{title_prefix}: CPU vs RAM vs GPU Active Utilization\n"
                 )
    ax.set_ylabel("Utilization (%)")
    ax.set_xticks(x_centres)
    ax.set_xticklabels(metrics)
    ax.legend()
    ax.grid(axis="y", alpha=0.3)
    plt.tight_layout()
    plt.show()

    # ----------------------------------------------------------
    # Graph 2 : Power Usage — one bar per scheme
    # ----------------------------------------------------------
    fig, ax = plt.subplots(figsize=(8, 6))

    bars = ax.bar(schemes, resource_data["Power"], color=COLORS)
    add_labels(ax, bars, "{:.0f}")

    ax.set_title(f"{title_prefix}: Average Power Consumption")
    ax.set_ylabel("Power (mW)")
    ax.grid(axis="y", alpha=0.3)
    plt.tight_layout()
    plt.show()

    # ----------------------------------------------------------
    # Graph 3 : FPS — one bar per scheme
    # ----------------------------------------------------------
    fig, ax = plt.subplots(figsize=(8, 6))

    bars = ax.bar(schemes, fps_data, color=COLORS)
    add_labels(ax, bars)

    ax.set_title(f"{title_prefix}: FPS Comparison")
    ax.set_ylabel("Frames Per Second")
    ax.grid(axis="y", alpha=0.3)
    plt.tight_layout()
    plt.show()

    # ----------------------------------------------------------
    # Graph 4 : Overall E2E Latency — grouped by PERCENTILE
    #           Each percentile cluster: Onboard | ALB | Offboard
    # ----------------------------------------------------------
    percentiles  = ["P50", "P90", "P99"]
    n_pctiles    = len(percentiles)
    x_centres_p  = np.arange(n_pctiles) * (n_schemes * bar_width + cluster_gap)

    fig, ax = plt.subplots(figsize=(10, 6))

    for i, (scheme, color) in enumerate(zip(schemes, COLORS)):
        values = [e2e_data[p][i] for p in percentiles]
        bars = ax.bar(x_centres_p + offsets[i], values, bar_width,
                      label=scheme, color=color)
        add_labels(ax, bars)

    ax.set_title(f"{title_prefix}: Overall End-to-End Latency\n"
                )
    ax.set_ylabel("Latency (ms)")
    ax.set_xticks(x_centres_p)
    ax.set_xticklabels(percentiles)
    ax.legend()
    ax.grid(axis="y", alpha=0.3)
    plt.tight_layout()
    plt.show()

    # ----------------------------------------------------------
    # Graph 5 : Application-wise Latency
    #           One subplot per app; within each subplot grouped
    #           by percentile with Onboard | ALB | Offboard bars
    # ----------------------------------------------------------
    apps = list(app_latency.keys())

    fig, axes = plt.subplots(1, 3, figsize=(18, 6))

    for ax, app in zip(axes, apps):
        x_c = np.arange(n_pctiles) * (n_schemes * bar_width + cluster_gap)

        for i, (scheme, color) in enumerate(zip(schemes, COLORS)):
            values = [app_latency[app][p][i] for p in percentiles]
            bars = ax.bar(x_c + offsets[i], values, bar_width,
                          label=scheme, color=color)
            add_labels(ax, bars)

        ax.set_title(app)
        ax.set_xticks(x_c)
        ax.set_xticklabels(percentiles)
        ax.set_ylabel("Latency (ms)")
        ax.grid(axis="y", alpha=0.3)

    handles, labels = axes[0].get_legend_handles_labels()
    fig.legend(handles, labels, loc="upper center", ncol=3)
    fig.suptitle(
        f"{title_prefix}: Application-wise Latency Comparison\n"
        f"(grouped by percentile — Onboard / ALB / Offboard per cluster)",
        fontsize=14, y=1.04
    )
    plt.tight_layout()
    plt.show()


# ============================================================
# NORMAL CONDITION PLOTS
# ============================================================

create_comparison_plots(
    "NORMAL CONDITION",
    normal_resource,
    normal_fps,
    normal_e2e,
    normal_app_latency
)

# ============================================================
# STRESS CONDITION PLOTS
# ============================================================

create_comparison_plots(
    "STRESS CONDITION",
    stress_resource,
    stress_fps,
    stress_e2e,
    stress_app_latency
)
```
