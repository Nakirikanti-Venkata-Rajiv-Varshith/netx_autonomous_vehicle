```python
import matplotlib.pyplot as plt
import numpy as np

# --------------------------------------------------
# ADD VALUES ON BARS
# --------------------------------------------------
def add_bar_labels(bars):
    for bar in bars:
        height = bar.get_height()

        plt.text(
            bar.get_x() + bar.get_width()/2,
            height,
            f'{height:.2f}',
            ha='center',
            va='bottom',
            rotation=0,
            fontsize=8
        )


# --------------------------------------------------
# PLOT GENERIC GROUPED BAR CHART
# --------------------------------------------------
def plot_metric(condition, metric_name, metric_data):

    labels = metric_data["labels"]

    offload = metric_data["OFFLOAD"]
    onboard = metric_data["ONBOARD"]
    alb = metric_data["ALB"]

    x = np.arange(len(labels))
    width = 0.25

    plt.figure(figsize=(10,6))

    bars1 = plt.bar(
        x - width,
        offload,
        width,
        label="OFFLOAD"
    )

    bars2 = plt.bar(
        x,
        onboard,
        width,
        label="ONBOARD"
    )

    bars3 = plt.bar(
        x + width,
        alb,
        width,
        label="ALB"
    )

    add_bar_labels(bars1)
    add_bar_labels(bars2)
    add_bar_labels(bars3)

    plt.xticks(x, labels)
    plt.ylabel("Value")
    plt.title(f"{metric_name} ({condition})")

    plt.legend()
    plt.grid(axis='y', linestyle='--', alpha=0.3)

    plt.tight_layout()
    plt.show()


# --------------------------------------------------
# RESOURCE UTILIZATION
# CPU RAM GPU ACTIVE_GPU
# POWER SEPARATED
# --------------------------------------------------
def plot_resource_utilization(condition):

    data_cond = data[condition]

    resource_labels = [
        "CPU %",
        "RAM %",
        "GPU %",
        "Active GPU %"
    ]

    offload = data_cond["Average Resource Utilization"]["OFFLOAD"][:4]
    onboard = data_cond["Average Resource Utilization"]["ONBOARD"][:4]
    alb = data_cond["Average Resource Utilization"]["ALB"][:4]

    x = np.arange(len(resource_labels))
    width = 0.25

    plt.figure(figsize=(10,6))

    bars1 = plt.bar(x-width, offload, width, label='OFFLOAD')
    bars2 = plt.bar(x, onboard, width, label='ONBOARD')
    bars3 = plt.bar(x+width, alb, width, label='ALB')

    add_bar_labels(bars1)
    add_bar_labels(bars2)
    add_bar_labels(bars3)

    plt.xticks(x, resource_labels)
    plt.ylabel("Utilization (%)")
    plt.title(f"Resource Utilization ({condition})")

    plt.legend()
    plt.grid(axis='y', linestyle='--', alpha=0.3)

    plt.tight_layout()
    plt.show()


# --------------------------------------------------
# POWER CONSUMPTION
# --------------------------------------------------
def plot_power(condition):

    data_cond = data[condition]

    labels = ["Power"]

    offload = [data_cond["Average Resource Utilization"]["OFFLOAD"][4]]
    onboard = [data_cond["Average Resource Utilization"]["ONBOARD"][4]]
    alb = [data_cond["Average Resource Utilization"]["ALB"][4]]

    x = np.arange(len(labels))
    width = 0.25

    plt.figure(figsize=(7,5))

    bars1 = plt.bar(x-width, offload, width, label='OFFLOAD')
    bars2 = plt.bar(x, onboard, width, label='ONBOARD')
    bars3 = plt.bar(x+width, alb, width, label='ALB')

    add_bar_labels(bars1)
    add_bar_labels(bars2)
    add_bar_labels(bars3)

    plt.xticks(x, labels)
    plt.ylabel("Power")
    plt.title(f"Power Consumption ({condition})")

    plt.legend()
    plt.grid(axis='y', linestyle='--', alpha=0.3)

    plt.tight_layout()
    plt.show()


# --------------------------------------------------
# GENERATE ALL FIGURES
# --------------------------------------------------

for condition in data:

    # Resource Utilization (without power)
    plot_resource_utilization(condition)

    # Power separate
    plot_power(condition)

    # Remaining metrics
    for metric_name, metric_data in data[condition].items():

        if metric_name == "Average Resource Utilization":
            continue

        plot_metric(
            condition,
            metric_name,
            metric_data
        )

print("All graphs generated successfully.")
```
